```python
import numpy as np

n = 100000

lancamentos = np.random.randint(0, 2, size=n)

num_caras = np.sum(lancamentos)
freq_relativa = num_caras / n

print("Número de caras:", num_caras)
print("Frequência relativa:", freq_relativa)
```
